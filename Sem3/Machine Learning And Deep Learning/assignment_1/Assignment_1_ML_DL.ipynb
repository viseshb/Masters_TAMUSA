```python
import numpy as np
import matplotlib.pyplot as plt

learning_rate = np.load('learning_rate.npy')
train_hist = np.load('train_hist.npy')
val_hist = np.load('val_hist.npy')

print(learning_rate)
print(train_hist)
print(val_hist)
```

```python
steps_per_epoch = 391
epochs = np.arange(1, 10)  # 1–9
epoch_markers = np.arange(steps_per_epoch, steps_per_epoch*10, steps_per_epoch)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# --- Subplot 1: Learning Rate ---
axes[0,0].plot(learning_rate, label="Learning Rate")
axes[0,0].scatter(epoch_markers, learning_rate[epoch_markers-1], c='blue', label="Epoch Markers")
axes[0,0].set_xlabel("Training Steps")
axes[0,0].set_ylabel("Learning Rate")
axes[0,0].tick_params(axis='x', rotation=30)
axes[0,0].legend()

# --- Subplot 2: Loss ---
axes[0,1].plot(epochs, train_hist[:,0], "-o", label="train")
axes[0,1].plot(epochs, val_hist[:,0], "-o", label="val")
axes[0,1].set_xlabel("Epochs")
axes[0,1].set_ylabel("Loss")
axes[0,1].tick_params(axis='x', rotation=30)
axes[0,1].legend()

# --- Subplot 3: Accuracy ---
axes[1,0].plot(epochs, train_hist[:,1], "-o", label="train")
axes[1,0].plot(epochs, val_hist[:,1], "-o", label="val")
axes[1,0].set_xlabel("Epochs")
axes[1,0].set_ylabel("Accuracy")
axes[1,0].tick_params(axis='x', rotation=30)
axes[1,0].legend()

# --- Subplot 4: F1 Score ---
axes[1,1].plot(epochs, train_hist[:,2], "-o", label="train")
axes[1,1].plot(epochs, val_hist[:,2], "-o", label="val")
axes[1,1].set_xlabel("Epochs")
axes[1,1].set_ylabel("F1 Score")
axes[1,1].tick_params(axis='x', rotation=30)
axes[1,1].legend()

plt.tight_layout()
plt.show()
```
