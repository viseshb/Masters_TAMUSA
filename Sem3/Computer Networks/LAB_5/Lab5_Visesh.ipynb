```python
# ===============================================================
#  Update OSPF Hello and Dead Intervals on All Nodes
# ===============================================================

nodes = {n.get_name(): n for n in slice.get_nodes()}
ospf_interfaces = ["enp7s0", "enp8s0"]  # Adjust if needed

for node_name, node in nodes.items():
    print(f"\n--- Updating OSPF config on {node_name} ---")

    try:
        # Build new ospfd.conf content
        ospf_conf = f"""hostname ospfd
password zebra
router ospf
 network 10.1.0.0/16 area 0
!
"""

        # Add interface-specific hello/dead interval settings
        for iface in ospf_interfaces:
            ospf_conf += f"""interface {iface}
 ip ospf hello-interval 2
 ip ospf dead-interval 5
!
"""

        # Upload config to /etc/quagga/ospfd.conf
        node.execute("sudo bash -c 'cat > /etc/quagga/ospfd.conf' <<'EOF'\n" + ospf_conf + "EOF")

        # Fix file ownership and permissions
        node.execute("sudo chown quagga:quagga /etc/quagga/ospfd.conf && sudo chmod 640 /etc/quagga/ospfd.conf")

        # Restart the ospfd daemon
        stdout, stderr = node.execute("sudo /etc/init.d/ospfd restart")
        print(stdout)

        # Optional: verify via vtysh CLI
        print(f"--- Verifying new OSPF timers on {node_name} ---")
        verify_cmd = "sudo vtysh -c 'show running-config' | grep -A2 interface"
        stdout, stderr = node.execute(verify_cmd)
        print(stdout)

    except Exception as e:
        print(f"Error updating {node_name}: {e}")
```

```python
# ===============================================================
# LAB 5 - OSPF Fault and Recovery Test (Sections 2.2 & 2.3)
# ===============================================================
import time
from datetime import datetime

# --- Select nodes ---
A = slice.get_node("ND_A")
B = slice.get_node("ND_B")

# --- Target IP (ND_C interface) ---
TARGET_IP = "10.1.2.2"
FAIL_IFACE = "enp7s0"

print(f"Starting OSPF fault-tolerance test from ND_A to {TARGET_IP}\n")

# --- Start continuous ping on ND_A in background ---
print("Starting continuous ping...")
A.execute(f"nohup ping -i 1 {TARGET_IP} > /tmp/ping_log.txt 2>&1 &", quiet=True)
time.sleep(3)

# --- Verify ping is running ---
print("Checking initial ping status (before link failure)...")
stdout, stderr = A.execute("tail -n 5 /tmp/ping_log.txt")
print(stdout)

# --- Simulate link failure on ND_B ---
print(f"\nSimulating link failure on {B.get_name()}:{FAIL_IFACE} ...")
B.execute(f"sudo ip link set dev {FAIL_IFACE} down")
fail_start = datetime.now()
print(f"Link brought DOWN at {fail_start.strftime('%H:%M:%S')}")

# --- Wait for OSPF reconvergence ---
print("Waiting 15 seconds for OSPF to detect failure and reroute...")
time.sleep(15)

# --- Check ping status after failure ---
print("\nPing results after link failure:")
stdout, stderr = A.execute("tail -n 10 /tmp/ping_log.txt")
print(stdout)

# --- Restore link ---
print(f"\nRestoring link on {B.get_name()}:{FAIL_IFACE} ...")
B.execute(f"sudo ip link set dev {FAIL_IFACE} up")
restore_time = datetime.now()
print(f"Link restored at {restore_time.strftime('%H:%M:%S')}")

# --- Allow OSPF to stabilize again ---
print("Waiting 10 seconds for OSPF to re-establish adjacency...")
time.sleep(10)

# --- Check ping after recovery ---
print("\nPing results after recovery:")
stdout, stderr = A.execute("tail -n 10 /tmp/ping_log.txt")
print(stdout)

# --- Display timing summary ---
print("Timing Summary:")
print(f" Link down: {fail_start.strftime('%H:%M:%S')}")
print(f" Link restored: {restore_time.strftime('%H:%M:%S')}")
print(" OSPF recovery expected approximately 5–7 seconds with new hello/dead timers.")
```
